```python
df = pd.read_csv("Sample _Superstore.csv", encoding="latin1")

df.head()
```

```python
df[["Order Date", "Sales", "Profit", "Region", "Category"]].head()
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])

df.info()
```

```python
df["Year"] = df["Order Date"].dt.year
df["Month"] = df["Order Date"].dt.month
df["YearMonth"] = df["Order Date"].dt.to_period("M").astype(str)
```

```python
df["Profit Margin"] = df["Profit"] / df["Sales"]
```

```python
df.head()
```

```python
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
profit_margin = total_profit / total_sales

print("Total Sales:", total_sales)
print("Total Profit:", total_profit)
print("Profit Margin:", profit_margin)
```

```python
# Insight:
# The company generates 2297200$ in sales with a profit margin of 12%.
```

```python
monthly_sales = df.groupby("YearMonth")["Sales"].sum().reset_index()

monthly_sales["YearMonth"] = pd.to_datetime(monthly_sales["YearMonth"])
monthly_sales = monthly_sales.sort_values("YearMonth")
```

```python
plt.figure(figsize=(12,6))
plt.plot(monthly_sales["YearMonth"], monthly_sales["Sales"])
plt.title("Monthly Sales Trend")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
# Insight:
# Sales show an overall upward trend with fluctuations across months.
# Some peaks may indicate seasonal demand or promotional periods.
```

```python
region_sales = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)
print(region_sales)
```

```python
region_sales.plot(kind="bar")
plt.title("Sales by Region")
plt.ylabel("Sales")
plt.show()
```

```python
# Insight:
# The highest sales come from West, while South underperforms.
# This might suggests potential for inefficiency in South.
```

```python
category_profit = df.groupby("Category")["Profit"].sum()
print(category_profit)
```

```python
category_profit.plot(kind="bar")
plt.title("Profit by Category")
plt.ylabel("Profit")
plt.show()
```

```python
# Insight:
# Some categories generate high revenue but lower profit,
# indicating cost or pricing inefficiencies.
```

```python
category_margin = df.groupby("Category")["Profit Margin"].mean()
print(category_margin)

category_margin.plot(kind="bar")
plt.title("Average Profit Margin by Category")
plt.ylabel("Profit Margin")
plt.show()
```

```python
# Insight:
# Furniture have significantly lower profit margins,
# indicating potential pricing or cost structure issues.
```

```python
subcat_profit = df.groupby("Sub-Category")["Profit"].sum().sort_values()
print(subcat_profit.head(10))

subcat_profit.head(10).plot(kind="bar")
plt.title("Top 10 Loss-Making Sub-Categories")
plt.ylabel("Profit")
plt.show()
```

```python
# Insight:
# Certain sub-categories consistently generate losses,
# which may require price adjustments or cost reduction.
```

```python
plt.scatter(df["Discount"], df["Profit"])
plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")
plt.show()
```

```python
# Insight:
# Higher discounts appear to be associated with lower profitability,
# suggesting that aggressive discounting negatively impacts margins.
```

```python
monthly_sales = df.groupby("YearMonth")["Sales"].sum().reset_index()
monthly_sales["YearMonth"] = pd.to_datetime(monthly_sales["YearMonth"])

# Convert to numeric time
monthly_sales["t"] = range(len(monthly_sales))

from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(monthly_sales[["t"]], monthly_sales["Sales"])

monthly_sales["Forecast"] = model.predict(monthly_sales[["t"]])

plt.figure(figsize=(12,6))
plt.plot(monthly_sales["YearMonth"], monthly_sales["Sales"], label="Actual")
plt.plot(monthly_sales["YearMonth"], monthly_sales["Forecast"], label="Forecast")
plt.legend()
plt.title("Sales Forecast")
plt.show()
```

```python
# Insight:
# The forecast indicates a general trend in sales development,
# which can support planning and budgeting decisions.
```
